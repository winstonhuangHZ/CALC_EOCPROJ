```python
#先跑一组200次实验，步长.0001, theta2从89.99到99.01
from pathlib import Path

init = 89.99
step = 0.0001

run_number = "01"

folder_path = Path(f"run{run_number}")
folder_path.mkdir(exist_ok=True)

param["theta2_0"] = init

for i in tqdm.tqdm(range(200)):
    current_theta2 = init + i * step
    param["theta2_0"] = current_theta2
    num = f"{i+1:04d}"
    t1_str = f"{param['theta1_0']:.4f}"
    t2_str = f"{current_theta2:.4f}"
    path = f"run{run_number}/run_{run_number}_data_{num}_{t1_str}_{t2_str}.parquet"
    #print(path)
    temp_storage = simulate_double_pendulum(param)
        df_out = pd.DataFrame(temp_storage, columns=['theta_1', 'theta_2', 'energy'])
        df_out.to_parquet(path, index=False)

```
